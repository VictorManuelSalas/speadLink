Nuevo Schema

Para poder subir tu nuevo schema a la base de datos alojado en el docker es necesario entrar al contenedor donde esta corriendo el backend.

    Para saber que contenedor es, seria ejecutando: docker ps
    Para entrar al contenedor es con: docker exec -it "name" sh
    Correr la migracion con: npx prisma migrate dev --name "nameDesc"

## Instalar nueva dependecia
docker compose run --rm api npm install "libreria"
